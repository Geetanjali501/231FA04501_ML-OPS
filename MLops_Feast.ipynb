```python
import pandas as pd
df = pd.read_csv("balanced_employability_dataset.csv")
print(df.head())
print(df.shape)
```

```python
print(df.columns)
print(df.info())
print(df.isnull().sum())
```

```python
import pandas as pd

df["student_id"] = df["Student_ID"].astype("int64")

df["cgpa"] = df["CGPA"].astype("float32")
df["attendance"] = df["Attendance"].astype("float32")

df["programming"] = df["Programming"].astype("float32")
df["data_structures"] = df["Data_Structures"].astype("float32")
df["algorithms"] = df["Algorithms"].astype("float32")
df["dbms"] = df["DBMS"].astype("float32")
df["operating_systems"] = df["Operating_Systems"].astype("float32")
df["computer_networks"] = df["Computer_Networks"].astype("float32")
df["oop"] = df["OOP"].astype("float32")
df["python"] = df["Python"].astype("float32")
df["java"] = df["Java"].astype("float32")
df["sql"] = df["SQL"].astype("float32")
df["web_development"] = df["Web_Development"].astype("float32")

df["git_github"] = df["Git_GitHub"].astype("float32")
df["cloud_computing"] = df["Cloud_Computing"].astype("float32")
df["ai_ml"] = df["AI_ML"].astype("float32")

df["communication"] = df["Communication"].astype("float32")
df["problem_solving"] = df["Problem_Solving"].astype("float32")
df["teamwork"] = df["Teamwork"].astype("float32")

df["internship"] = df["Internship"].astype("int64")
df["projects"] = df["Projects"].astype("int64")
df["certifications"] = df["Certifications"].astype("int64")
df["aptitude"] = df["Aptitude"].astype("float32")

df["average_skill"] = df["Average_Skill"].astype("float32")
df["industry_requirement"] = df["Industry_Requirement"].astype("float32")
df["skill_gap"] = df["Skill_Gap"].astype("float32")

df["Gender"] = df["Gender"].astype(str).str.strip().str.lower()

df["gender_encoded"] = df["Gender"].map({
    "male": 0,
    "female": 1
})

df["gender_encoded"] = df["gender_encoded"].fillna(0).astype("int64")

df["employability"] = df["Employability"].astype("int64")

print(df.head())
```

```python
df[
    [
        "student_id",
        "gender_encoded",
        "cgpa",
        "attendance",
        "programming",
        "data_structures",
        "algorithms",
        "dbms",
        "operating_systems",
        "computer_networks",
        "oop",
        "python",
        "java",
        "sql",
        "web_development",
        "git_github",
        "cloud_computing",
        "ai_ml",
        "communication",
        "problem_solving",
        "teamwork",
        "internship",
        "projects",
        "certifications",
        "aptitude",
        "average_skill",
        "industry_requirement",
        "skill_gap",
        "employability"
    ]
].head()
```

```python
base_time = pd.Timestamp(
    "2026-01-01",
    tz="UTC"
)

df["event_timestamp"] = (
    base_time +
    pd.to_timedelta(
        df["student_id"],
        unit="s"
    )
)

df["created_timestamp"] = (
    df["event_timestamp"] +
    pd.Timedelta(seconds=1)
)
```

```python
feature_df = df[
    [
        "student_id",
        "event_timestamp",
        "created_timestamp",
        "gender_encoded",
        "cgpa",
        "attendance",
        "programming",
        "data_structures",
        "algorithms",
        "dbms",
        "operating_systems",
        "computer_networks",
        "oop",
        "python",
        "java",
        "sql",
        "web_development",
        "git_github",
        "cloud_computing",
        "ai_ml",
        "communication",
        "problem_solving",
        "teamwork",
        "internship",
        "projects",
        "certifications",
        "aptitude",
        "average_skill",
        "industry_requirement",
        "skill_gap"
    ]
].copy()

label_df = df[
    [
        "student_id",
        "event_timestamp",
        "employability"
    ]
].copy()
```

```python
print("Feature data:")
display(feature_df.head())
print("Labels:")
display(label_df.head())
```

```python
import os

repo_path = "/content/employability_feast"

os.makedirs(
    f"{repo_path}/data",
    exist_ok=True
)
```

```python
feature_df.to_parquet(
    f"{repo_path}/data/employability_features.parquet",
    index=False
)
```

```python
feature_store_yaml = """
project: employability_project

registry: data/registry.db

provider: local

offline_store:
  type: file

online_store:
  type: sqlite
  path: data/online_store.db
"""

with open(
    f"{repo_path}/feature_store.yaml",
    "w"
) as f:
    f.write(feature_store_yaml)
```

```python
feature_definition = '''
from datetime import timedelta

from feast import (
    Entity,
    FeatureView,
    FeatureService,
    Field,
    FileSource
)

from feast.types import (
    Float32,
    Int64
)

student = Entity(
    name="student",
    join_keys=["student_id"],
    description="CSE student"
)

employability_source = FileSource(
    name="employability_source",
    path="data/employability_features.parquet",
    timestamp_field="event_timestamp",
    created_timestamp_column="created_timestamp"
)

employability_feature_view = FeatureView(
    name="employability_features",
    entities=[student],

    ttl=timedelta(days=50000),

    schema=[
        Field(name="gender_encoded", dtype=Int64),
        Field(name="cgpa", dtype=Float32),
        Field(name="attendance", dtype=Float32),
        Field(name="programming", dtype=Float32),
        Field(name="data_structures", dtype=Float32),
        Field(name="algorithms", dtype=Float32),
        Field(name="dbms", dtype=Float32),
        Field(name="operating_systems", dtype=Float32),
        Field(name="computer_networks", dtype=Float32),
        Field(name="oop", dtype=Float32),
        Field(name="python", dtype=Float32),
        Field(name="java", dtype=Float32),
        Field(name="sql", dtype=Float32),
        Field(name="web_development", dtype=Float32),
        Field(name="git_github", dtype=Float32),
        Field(name="cloud_computing", dtype=Float32),
        Field(name="ai_ml", dtype=Float32),
        Field(name="communication", dtype=Float32),
        Field(name="problem_solving", dtype=Float32),
        Field(name="teamwork", dtype=Float32),
        Field(name="internship", dtype=Int64),
        Field(name="projects", dtype=Int64),
        Field(name="certifications", dtype=Int64),
        Field(name="aptitude", dtype=Float32),
        Field(name="average_skill", dtype=Float32),
        Field(name="industry_requirement", dtype=Float32),
        Field(name="skill_gap", dtype=Float32)
    ],

    source=employability_source,

    online=True
)

employability_feature_service = FeatureService(
    name="employability_prediction_service",
    features=[
        employability_feature_view
    ]
)
'''

with open(
    f"{repo_path}/features.py",
    "w"
) as f:
    f.write(feature_definition)
```
